```python
from elasticsearch import Elasticsearch
```

```python
es = Elasticsearch(
    "http://localhost:9200",
    basic_auth=("elastic", "changeme") 
)
```

```python
es.indices.create(index="my_index", ignore=400)
```

```python
doc = {
    "author": "kimchy",
    "text": "Elasticsearch: cool. bonsai cool.",
    "timestamp": "2024-01-24T14:12:12",
}

es.index(index="my_index", id=1, document=doc)
```

```python
 resp = es.search(index="my_index", query={"query_string":{
                "query": "bonsai",
                "fields": ["text"], #fields you want to search
                "default_operator": "OR"
            }})
resp
```
